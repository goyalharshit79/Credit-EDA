### Lets try to find out if there is any disparity in the number of clients based on various factors

```python
#lets define two functions

#1 - that plots the count of how many times that particular value occurs

def plot_value_count(col, ax = None):
    return sns.barplot(x = app_data_sample[col].value_counts().index, y = app_data_sample[col].value_counts(), ax = ax)

#2 - that plots the total amount of credit disposed to that particular sub-category

def plot_amt_credit(col, ax = None):
    grouped_data = app_data_sample.groupby(col)["AMT_CREDIT"].sum()
    return sns.barplot(x = grouped_data.index, y = grouped_data, ax = ax)
```

### Gender

```python
figure,ax = plt.subplots(nrows = 1, ncols = 2, figsize = (20,5))

plot_value_count("CODE_GENDER", ax = ax[0])
ax[0].set_title("By Count")

plot_amt_credit("CODE_GENDER", ax = ax[1])
ax[1].set_title("By Amount")

plt.show()
```

#### Both the number of credits and the amount of credit dispensed seems to be almost double in case of Females as compared to males

### Marital Status

```python
figure,ax = plt.subplots(nrows = 1, ncols = 2, figsize = (20,5))

plot_value_count("NAME_FAMILY_STATUS", ax = ax[0])
ax[0].set_title("By Count")

plot_amt_credit("NAME_FAMILY_STATUS", ax = ax[1])
ax[1].set_title("By Amount")

plt.show()
```

#### The number of married people that have applied for a loan is much higher than those that are unmarried.
- I am going to assume that the larger amount of type of loans among married clientelle is going to be house loans.

### Now lets see of a higher number of previous applications translates into having less difficulties in paying back the loan

#### I am going to set the threshold of previous loans at 4, because,
- That is our median value
- Moreover, 4 previous loans is a good enough indicator of credit worthiness, where it will highlight if there are patterns of default in the client

```python
by_count_currid_df = by_count_currid.to_frame()
by_count_currid_df.head()
```

```python
#Function to get the targets from the dataset
def get_targets(curr_id):
    return app_data_sample[app_data_sample["SK_ID_CURR"] == curr_id]["TARGET"].values[0]
```

```python
by_count_currid_df = by_count_currid_df.reset_index()
```

```python
by_count_currid_df.head()
```

```python
by_count_currid_df["TARGET"] = by_count_currid_df["SK_ID_CURR"].apply(lambda x: get_targets(x))
by_count_currid_df.shape
```

```python
by_count_currid_df.head()
```

```python
#grouping using number of prev ids to see if the number of prev loans has any bearing on the difficulty to pay current loans
by_prev = by_count_currid_df.groupby("SK_ID_PREV")["TARGET"].sum()
```

```python
plt.figure(figsize =(20,5))
by_prev.plot.bar()
```

#### Here the higher the bar, means more 1s in the Target column, which means more instances of difficulties in payements.

- Now I understand that as we go towards the higher number of previous loans, the amount of data for them also decreases.
- But if we consider the 50th per centile as a boundary, which is 4, then we can see a stark decrease in the sizes of bars.

#### This implies that having more successfull loans in the past can also translate into having more chances of being credit worthy. And this is relatable to real life scenario as well. We see that the more successfull we are at past credits, helps us build our credit score, hence more credit worthiness

```python
#Total number of 1s in target column where there are more than 4 instances of loan
by_count_currid_df[by_count_currid_df["SK_ID_PREV"] > 4]["TARGET"].sum()
```
